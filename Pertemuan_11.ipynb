Coin Change Greedy

```python
def coin_change_greedy(n):
    coins = [20, 10, 5, 1]
    i = 0

    while(n > 0):
        if(coins[i] > n):
            i = i + 1
        else:
            print(coins[i])
            n = n - coins[i]
    print("\n\n")

coin_change_greedy(57)
```

Activity Selection (Greedy)

```python
# n --> Jumlah total kegiatan
# s[] --> Array yang berisi waktu mulai semua aktivitas
# f[] --> Array yang berisi waktu selesai semua aktivitas

def printMaxActivities(s, f):
    n = len(f)
    print("The following activities are selected")

    # aktivitas pertama selalu terpilih duluan
    i = 0
    print(i),

    # Pertimbangkan aktivitas lainnya
    for j in range(n):

        # Jika aktivitas ini memiliki waktu mulai lebih besar
        # atau sama dengan waktu selesai aktivitas yang dipilih sebelumnya,
        # maka pilih aktivitas tersebut
        if s[j] >= f[i]:
            print(j),
            i = j

# Program untuk cek pemanggilan fungsi di atas
s = [1, 3, 0, 5, 8, 5]
f = [2, 4, 6, 7, 9, 9]
printMaxActivities(s, f)
```

0/1 Knapsack (Dynamic Programming)

```python
def knapsack(wt, val, W, n):

    # base conditions
    if n == 0 or W == 0:
        return 0
    if t[n][W] != -1:
        return t[n][W]

    # choice diagram code
    if wt[n-1] <= W:
        t[n][W] = max(
            val[n-1] + knapsack(
                wt, val, W-wt[n-1], n-1),
            knapsack(wt, val, W, n-1))
        return t[n][W]
    elif wt[n-1] > W:
        t[n][W] = knapsack(wt, val, W, n-1)
        return t[n][W]

# Driver code
if __name__ == '__main__':
    profit = [60, 100, 120]
    weight = [10, 20, 30]
    W = 50
    n = len(profit)

    # We initialize the matrix with -1 at first.
    t = [[-1 for i in range(W + 1)] for j in range(n + 1)]
    print("Maximum Value that can be Obtained is", knapsack(weight, profit, W, n))
```

Fractional Knapsack (Greedy)

```python
class Item:
    def __init__(self, weight, value):
        self.weight = weight
        self.value = value
        # Calculate the value-to-weight ratio for each item
        self.ratio = value / weight

def fractional_knapsack(items, capacity):
    # Sort items by their value-to-weight ratio in non-increasing order
    items.sort(key=lambda x: x.ratio, reverse=True)

    total_value = 0

    # Initialize remaining capacity of the knapsack
    remaining_capacity = capacity

    # Iterate through the sorted items list
    for item in items:
        if remaining_capacity <= 0:
            break

        # Calculate fraction of the item that fits into the knapsack
        fraction = min(1, remaining_capacity / item.weight)

        # Update total value and remaining capacity
        total_value += fraction * item.value
        remaining_capacity -= fraction * item.weight

    # Return total maximum value obtained
    return round(total_value, 2)

# Example usage
items = [Item(10, 60), Item(20, 100), Item(30, 120)]
capacity = 50
print("Maximum Value that can be Obtained is", fractional_knapsack(items, capacity))
```

Activity Selection dengan Data Dictionary (Greedy)

```python
data = {
    "start_time": [2, 6, 4, 10, 13, 7],
    "finish_time": [5, 10, 8, 12, 14, 15],
    "activity": ["Homework", "Presentation", "Term paper",
                 "Volleyball practice", "Biology lecture", "Hangout"]
}

selected_activity = []
start_position = 0

# Sorting the items in ascending order with respect to finish time
tem = 0
for i in range(0, len(data['finish_time'])):
    for j in range(0, len(data['finish_time'])):
        if data['finish_time'][i] < data['finish_time'][j]:
            tem = data['activity'][i]
            data['activity'][i] = data['activity'][j]
            data['activity'][j] = tem

            tem = data['finish_time'][i]
            data['finish_time'][i] = data['finish_time'][j]
            data['finish_time'][j] = tem

            tem = data['start_time'][i]
            data['start_time'][i] = data['start_time'][j]
            data['start_time'][j] = tem

# By default, the first activity is inserted in the list of activities to be selected.
selected_activity.append(data['activity'][start_position])

for pos in range(len(data['finish_time'])):
    if data['start_time'][pos] >= data['finish_time'][start_position]:
        selected_activity.append(data['activity'][pos])
        start_position = pos

print(f"The student can work on the following activities: {selected_activity}")

# Results
# The student can work on the following activities:
# ['Homework', 'Presentation', 'Volleyball practice', 'Biology lecture']
```

Masalah Pengepakan Sampah (Bin Packing — First Fit)

```python
def bin_packing(berat, kapasitas):
    n = len(berat)
    # Simpan sisa kapasitas tiap bin
    bin_sisa = []

    for item in berat:
        ditempatkan = False

        # Cari bin pertama yang muat (First Fit)
        for i in range(len(bin_sisa)):
            if bin_sisa[i] >= item:
                bin_sisa[i] -= item
                ditempatkan = True
                break

        # Jika tidak ada bin yang muat, buka bin baru
        if not ditempatkan:
            bin_sisa.append(kapasitas - item)

    return len(bin_sisa)


# Contoh 1
berat1 = [4, 8, 1, 4, 2, 1]
kapasitas1 = 10
hasil1 = bin_packing(berat1, kapasitas1)
print("=== Contoh 1 ===")
print("berat[]  :", berat1)
print("Kapasitas:", kapasitas1)
print("Keluaran :", hasil1)
print("Dibutuhkan minimal", hasil1, "tempat sampah untuk menampung semua barang\n")

# Contoh 2
berat2 = [9, 8, 2, 2, 5, 4]
kapasitas2 = 10
hasil2 = bin_packing(berat2, kapasitas2)
print("=== Contoh 2 ===")
print("berat[]  :", berat2)
print("Kapasitas:", kapasitas2)
print("Keluaran :", hasil2)
print("Dibutuhkan minimal", hasil2, "tempat sampah untuk menampung semua barang")
```

Polisi dan Pencuri (Greedy)

```python
def tangkap_pencuri(arr, k):
    n = len(arr)

    # Pisahkan indeks polisi dan pencuri
    polisi = []
    pencuri = []

    for i in range(n):
        if arr[i] == 'P':
            polisi.append(i)
        elif arr[i] == 'T':
            pencuri.append(i)

    # Greedy: cocokkan polisi dan pencuri terdekat
    p = 0  # pointer polisi
    t = 0  # pointer pencuri
    tangkapan = 0

    while p < len(polisi) and t < len(pencuri):
        # Jika jarak polisi ke pencuri <= k, tangkap
        if abs(polisi[p] - pencuri[t]) <= k:
            tangkapan += 1
            p += 1
            t += 1
        # Jika pencuri terlalu jauh di kiri, geser pointer pencuri
        elif polisi[p] > pencuri[t]:
            t += 1
        # Jika polisi terlalu jauh di kiri, geser pointer polisi
        else:
            p += 1

    return tangkapan


# Contoh 1
arr1 = ['P', 'T', 'T', 'P', 'T']
k1 = 1
print("=== Contoh 1 ===")
print("arr[] =", arr1, ", k =", k1)
print("Keluaran:", tangkap_pencuri(arr1, k1))
# Output: 2

# Contoh 2
arr2 = ['T', 'T', 'P', 'P', 'T', 'P']
k2 = 2
print("\n=== Contoh 2 ===")
print("arr[] =", arr2, ", k =", k2)
print("Keluaran:", tangkap_pencuri(arr2, k2))
# Output: 3

# Contoh 3
arr3 = ['P', 'T', 'P', 'T', 'T', 'P']
k3 = 3
print("\n=== Contoh 3 ===")
print("arr[] =", arr3, ", k =", k3)
print("Keluaran:", tangkap_pencuri(arr3, k3))
# Output: 3
```
